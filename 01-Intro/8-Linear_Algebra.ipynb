```python
u = np.array([1,2,3])
v = np.array([4,5,6])
```

```python
u + v
```

```python
# multiplication 
u * v
```

```python
# dot product
def vector_multiplication(u,v):
    assert u.shape[0] == v.shape[0], "Vectors must be the same shape"
    n = u.shape[0]
    result = 0
    for i in range(n):
        result = result + u[i] * v[i]
    return result
```

```python
vector_multiplication(u,v)
```

```python
u.dot(v)
```

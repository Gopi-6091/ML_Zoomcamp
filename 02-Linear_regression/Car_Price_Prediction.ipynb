```python
base = ['engine_hp','engine_cylinders','highway_mpg','city_mpg','popularity']

df_train[base]
```

```python
X_train = df_train[base].values
```

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    return w_full[0], w_full[1:]
```

```python
X_train = df_train[base].fillna(0).values
X_train
```

```python
w0, w = train_linear_regression(X_train, y_train)
```

```python
y_pred = w0 + X_train.dot(w)
```

```python
# 2.9 RMSE
def rmse(y , y_pred):
    error = y - y_pred
    se = error ** 2
    mse = se.mean()
    return np.sqrt(mse)
```

```python
# all categorical variables

categorical_variables = ['make', 'model', 'engine_fuel_type', 'transmission_type', 'driven_wheels', 'market_category', 'vehicle_size', 'vehicle_style']
```

```python
categories = {}
for c in categorical_variables:
    categories[c] = list(df[c].value_counts().head().index)
```

```python
categories
```

```python
def prepare_X(df):
    df = df.copy()

    df['age'] = 2017 - df.year
    features = base + ['age']

    for v in [2, 3, 4]:
        df['num_of_doors_%s' % v] = (df.number_of_doors == v).astype('int')
        features.append('num_of_doors_%s' % v)
    for v in makes:
        df['make_%s' % v] = (df.make == v).astype('int')
        features.append('make_%s' % v)
    for c, values in categories.items():
        for v in values:
            df['%s_%s' % (c,v)] = (df[c] == v).astype('int')
            features.append('%s_%s' % (c,v))
        
    df_num = df[features]
    df_num = df_num.fillna(0)
    X= df_num.values
    return X
```

```python
X_train = prepare_X(df_train)
w0, w = train_linear_regression(X_train, y_train)

X_val = prepare_X(df_val)
y_pred = w0 + X_val.dot(w)

rmse(y_val, y_pred)
```

```python
# 2.13 Regularization
def train_linear_regression_reg(X, y, r=0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])

    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    return w_full[0], w_full[1:]
```

```python
X_train = prepare_X(df_train)
w0, w = train_linear_regression_reg(X_train, y_train, r=0.01)

X_val = prepare_X(df_val)
y_pred = w0 + X_val.dot(w)
rmse(y_val, y_pred)
```

```python
# 2.14 finding best paraments tuning the model

for r in [0.0 , 0.00001, 0.0001 , 0.001 , 0.01, 1 , 10]:
    X_train = prepare_X(df_train)
    w0, w = train_linear_regression_reg(X_train, y_train, r=r)

    X_val = prepare_X(df_val)
    y_pred = w0 + X_val.dot(w)
    score = rmse(y_val, y_pred)

    print(r,w0,score)
```
